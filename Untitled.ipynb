```python
import numpy as np
import pandas as pd

# -----------------------------------------------------------------------------
# 1. Configuration & Constants
# -----------------------------------------------------------------------------
rng = np.random.default_rng(42)

N_TARGET_ROWS = 20_000
INTERVAL_MIN = 15  # minutes between readings
HORIZON_HOURS = 48  # predict 2 days ahead
HORIZON_STEPS = int(HORIZON_HOURS * 60 / INTERVAL_MIN)
WARMUP_HOURS = 24  # history needed for 24h rolling features
WARMUP_STEPS = int(WARMUP_HOURS * 60 / INTERVAL_MIN)

N_TOTAL = N_TARGET_ROWS + HORIZON_STEPS + WARMUP_STEPS
START_DATE = pd.Timestamp("2025-01-01 00:00:00")

# Sensor ranges & calibration values
TEMP_MIN, TEMP_MAX = 20.5, 42.0  # DHT11, deg C
HUMIDITY_MIN, HUMIDITY_MAX = 40.0, 100.0  # DHT11, %
RAIN_MIN, RAIN_MAX = 0, 4095  # analog, dry=4095, wet=0

# Physically realistic atmospheric pressure envelope (the true underlying
# signal, before any per-sensor glitching is applied)
PRESSURE_REALISTIC_MEAN = 99_500  # Pa baseline
PRESSURE_REALISTIC_MIN, PRESSURE_REALISTIC_MAX = 95_500, 103_500

# Per-sensor hardware limits, taken from your actual observed extremes.
# Each sensor gets its own glitch range instead of a single shared one.
BMP180_HW_MIN, BMP180_HW_MAX = 66_592, 252_723
BMP280_HW_MIN, BMP280_HW_MAX = 71_322, 121_127
PRESSURE_SPIKE_PROB = 0.0015  # chance of extreme glitch reading, per sensor

# Mean-reversion (Ornstein-Uhlenbeck) parameters for the pressure "weather
# system" driver. A real high/low pressure system persists for a few days
# then decays — it does NOT drift forever like an unbounded random walk.
# We pick a ~2-day half-life and a stationary std of ~400 Pa, both typical
# of real synoptic pressure swings.
OU_HALFLIFE_STEPS = int(48 * 60 / INTERVAL_MIN)  # 2 days
OU_THETA = 1 - 0.5 ** (1 / OU_HALFLIFE_STEPS)      # per-step mean-reversion rate
OU_STATIONARY_STD = 400.0                          # Pa, long-run swing size
OU_SIGMA = OU_STATIONARY_STD * np.sqrt(2 * OU_THETA)  # per-step noise std

# Soil moisture (12-bit analog, inverted: dry ≈ 4095, wet ≈ 1000)
SOIL_MOISTURE_DRY = 4095
SOIL_MOISTURE_WET = 1000

# -----------------------------------------------------------------------------
# 2. Base Signals & Timeline
# -----------------------------------------------------------------------------
timestamps = pd.date_range(START_DATE, periods=N_TOTAL, freq=f"{INTERVAL_MIN}min")
hour = timestamps.hour + timestamps.minute / 60.0
doy = timestamps.dayofyear.values.astype(float)

hour_sin = np.sin(2 * np.pi * hour / 24)
hour_cos = np.cos(2 * np.pi * hour / 24)
doy_sin = np.sin(2 * np.pi * doy / 365.25)
doy_cos = np.cos(2 * np.pi * doy / 365.25)

seasonal_wetness = (np.sin(2 * np.pi * (doy - 105) / 365.25) + 1) / 2
diurnal_temp = np.sin(2 * np.pi * (hour - 9) / 24)
diurnal_humidity = -np.sin(2 * np.pi * (hour - 9) / 24)

# Temperature & Humidity (DHT11)
temp_dht11 = np.clip(
    30 - 5 * seasonal_wetness + 5 * diurnal_temp + rng.normal(0, 0.6, N_TOTAL),
    TEMP_MIN,
    TEMP_MAX,
)
humidity = np.clip(
    55
    + 30 * seasonal_wetness
    + 12 * diurnal_humidity
    + rng.normal(0, 2.5, N_TOTAL),
    HUMIDITY_MIN,
    HUMIDITY_MAX,
)

# -----------------------------------------------------------------------------
# 3. Barometric Pressure (mean-reverting "weather system" driver) & Glitches
# -----------------------------------------------------------------------------
# Mean-reverting driver instead of an unbounded random walk: each step decays
# toward 0 by OU_THETA and gets a fresh noise kick. This keeps pressure
# swings bounded and gives the 48h-ahead value real dependence on the
# current trend (learnable), rather than accumulating unrelated noise
# forever (unlearnable).
ou_walk = np.zeros(N_TOTAL)
ou_noise = rng.normal(0, OU_SIGMA, N_TOTAL)
for i in range(1, N_TOTAL):
    ou_walk[i] = ou_walk[i - 1] * (1 - OU_THETA) + ou_noise[i]

pressure_base = np.clip(
    PRESSURE_REALISTIC_MEAN - 700 * seasonal_wetness + ou_walk,
    PRESSURE_REALISTIC_MIN,
    PRESSURE_REALISTIC_MAX,
)


def inject_spikes(arr, hw_min, hw_max):
    """Randomly replace a small fraction of readings with hardware glitch
    values drawn from this sensor's own observed extreme range."""
    arr = arr.copy()
    spike_mask = rng.random(N_TOTAL) < PRESSURE_SPIKE_PROB
    arr[spike_mask] = rng.uniform(hw_min, hw_max, spike_mask.sum())
    return arr


pressure_bmp180 = np.clip(
    inject_spikes(pressure_base + rng.normal(0, 40, N_TOTAL), BMP180_HW_MIN, BMP180_HW_MAX),
    BMP180_HW_MIN,
    BMP180_HW_MAX,
)
pressure_bmp280 = np.clip(
    inject_spikes(pressure_base - 60 + rng.normal(0, 40, N_TOTAL), BMP280_HW_MIN, BMP280_HW_MAX),
    BMP280_HW_MIN,
    BMP280_HW_MAX,
)

temp_bmp180 = temp_dht11 + 1.2 + rng.normal(0, 0.8, N_TOTAL)
temp_bmp280 = temp_dht11 + 0.5 + rng.normal(0, 0.8, N_TOTAL)

# -----------------------------------------------------------------------------
# 4. Rain & Soil Moisture Simulation
# -----------------------------------------------------------------------------
pressure_trend_6h_base = (
    pd.Series(pressure_base)
    .diff(int(6 * 60 / INTERVAL_MIN))
    .fillna(0)
    .values
)
driver = (
    0.035 * (humidity - 70)
    + 2.2 * seasonal_wetness
    - 0.004 * pressure_trend_6h_base
    + rng.normal(0, 0.5, N_TOTAL)
)
driver_smooth = (
    pd.Series(driver).rolling(4, min_periods=1, center=True).mean().values
)

rain_prob = 1 / (1 + np.exp(-(driver_smooth - 1.0)))
is_raining = rng.random(N_TOTAL) < rain_prob
rain_intensity = np.where(is_raining, np.clip(driver_smooth - 0.5, 0, None), 0.0)
rain_intensity = rain_intensity / (rain_intensity.max() + 1e-9)

rain_raw = np.clip(
    RAIN_MAX - rain_intensity * RAIN_MAX + rng.normal(0, 60, N_TOTAL),
    RAIN_MIN,
    RAIN_MAX,
)

# Soil moisture leaky-bucket (0.0 = bone dry, 1.0 = fully saturated)
soil_wetness = np.zeros(N_TOTAL)
soil_wetness[0] = 0.3

for i in range(1, N_TOTAL):
    dry_loss = soil_wetness[i - 1] * 0.004
    rain_gain = rain_intensity[i] * 0.55
    soil_wetness[i] = np.clip(
        soil_wetness[i - 1] - dry_loss + rain_gain, 0.0, 1.0
    )

# Map wetness (0..1) to inverted sensor scale (4095..1000)
soil_moisture_raw = SOIL_MOISTURE_DRY - soil_wetness * (
    SOIL_MOISTURE_DRY - SOIL_MOISTURE_WET
)
soil_moisture = np.clip(
    soil_moisture_raw + rng.normal(0, 15, N_TOTAL),
    SOIL_MOISTURE_WET,
    SOIL_MOISTURE_DRY,
)

# -----------------------------------------------------------------------------
# 5. DataFrame Assembly & Feature Engineering
# -----------------------------------------------------------------------------
df = pd.DataFrame(
    {
        "timestamp": timestamps,
        "hour_sin": hour_sin,
        "hour_cos": hour_cos,
        "doy_sin": doy_sin,
        "doy_cos": doy_cos,
        "month": timestamps.month,
        "temperature": temp_dht11.round(1),
        "humidity": humidity.round(1),
        "temp_bmp180": temp_bmp180.round(1),
        "pressure_loc1": pressure_bmp180.round(0).astype(int),
        "temp_bmp280": temp_bmp280.round(1),
        "pressure_loc2": pressure_bmp280.round(0).astype(int),
        "rain_raw": rain_raw.round(0).astype(int),
        "soil_moisture": soil_moisture.round(0).astype(int),
    }
)

STEPS_3H = int(3 * 60 / INTERVAL_MIN)
STEPS_6H = int(6 * 60 / INTERVAL_MIN)
STEPS_24H = int(24 * 60 / INTERVAL_MIN)

df["pressure_loc1_trend_3h"] = df["pressure_loc1"].diff(STEPS_3H)
df["pressure_loc1_trend_6h"] = df["pressure_loc1"].diff(STEPS_6H)
df["pressure_loc2_trend_3h"] = df["pressure_loc2"].diff(STEPS_3H)
df["pressure_loc2_trend_6h"] = df["pressure_loc2"].diff(STEPS_6H)
df["pressure_diff"] = df["pressure_loc1"] - df["pressure_loc2"]
df["humidity_trend_3h"] = df["humidity"].diff(STEPS_3H)
df["rain_rate"] = df["rain_raw"].diff(1)

rain_wetness = (RAIN_MAX - df["rain_raw"]) / RAIN_MAX
df["rain_accum_6h"] = rain_wetness.rolling(STEPS_6H, min_periods=1).sum()
df["rain_accum_24h"] = rain_wetness.rolling(STEPS_24H, min_periods=1).sum()

# -----------------------------------------------------------------------------
# 6. Target Generation (48h forward shift)
# -----------------------------------------------------------------------------
df["temperature_target"] = df["temperature"].shift(-HORIZON_STEPS)
df["humidity_target"] = df["humidity"].shift(-HORIZON_STEPS)
df["pressure_loc1_target"] = df["pressure_loc1"].shift(-HORIZON_STEPS)
df["pressure_loc2_target"] = df["pressure_loc2"].shift(-HORIZON_STEPS)


# Categorize future rain
def rain_to_category(raw):
    if raw >= 3200:
        return "dry"
    elif raw >= 2200:
        return "light"
    elif raw >= 1000:
        return "moderate"
    else:
        return "heavy"


df["rain_category_target"] = (
    df["rain_raw"]
    .shift(-HORIZON_STEPS)
    .apply(lambda v: rain_to_category(v) if pd.notna(v) else np.nan)
)

# Flood risk calculation (using inverted soil moisture)
rain_accum_24h_norm = df["rain_accum_24h"] / df["rain_accum_24h"].max()
soil_wetness_target = (
    SOIL_MOISTURE_DRY - df["soil_moisture"].shift(-HORIZON_STEPS)
) / (SOIL_MOISTURE_DRY - SOIL_MOISTURE_WET)
flood_score = (
    0.6 * rain_accum_24h_norm.shift(-HORIZON_STEPS) + 0.4 * soil_wetness_target
)


def score_to_risk(score):
    if pd.isna(score):
        return np.nan
    elif score < 0.35:
        return "low"
    elif score < 0.65:
        return "watch"
    else:
        return "high"


df["flood_risk_target"] = flood_score.apply(score_to_risk)

# -----------------------------------------------------------------------------
# 7. Trim Warm-up, Order Columns & Export
# -----------------------------------------------------------------------------
df_final = df.iloc[WARMUP_STEPS : WARMUP_STEPS + N_TARGET_ROWS].reset_index(
    drop=True
)

COLUMN_ORDER = [
    "timestamp",
    "hour_sin",
    "hour_cos",
    "doy_sin",
    "doy_cos",
    "month",
    "temperature",
    "humidity",
    "temp_bmp180",
    "pressure_loc1",
    "temp_bmp280",
    "pressure_loc2",
    "rain_raw",
    "soil_moisture",
    "pressure_loc1_trend_3h",
    "pressure_loc1_trend_6h",
    "pressure_loc2_trend_3h",
    "pressure_loc2_trend_6h",
    "pressure_diff",
    "humidity_trend_3h",
    "rain_rate",
    "rain_accum_6h",
    "rain_accum_24h",
    "temperature_target",
    "humidity_target",
    "pressure_loc1_target",
    "pressure_loc2_target",
    "rain_category_target",
    "flood_risk_target",
]

df_final = df_final[COLUMN_ORDER]
df_final.to_csv("weather_flood_dataset.csv", index=False)
print(f"Done! Generated exactly {len(df_final):,} rows to 'weather_flood_dataset.csv'.")
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("weather_flood_dataset.csv")

print("=== 1. STRUCTURAL CHECKS ===")
print(f"Total Rows: {len(df)} (Expected: 20,000)")
print(f"Total Missing Values: {df.isna().sum().sum()} (Expected: 0)")

print("\n=== 2. SENSOR RANGE & POLARITY CHECKS ===")
# Soil moisture check (inverted 12-bit: dry should be higher than wet)
print(f"Soil Moisture Min: {df['soil_moisture'].min()} (Floor around ~1000 for wet)")
print(f"Soil Moisture Max: {df['soil_moisture'].max()} (Ceiling around ~4095 for dry)")

# Rain raw check (inverted: dry ≈ 4095)
print(f"Rain Raw Min: {df['rain_raw'].min()} | Max: {df['rain_raw'].max()}")

print("\n=== 3. TARGET CLASS DISTRIBUTION ===")
print("Rain Categories (48h Ahead):")
print(df["rain_category_target"].value_counts(normalize=True).round(3))

print("\nFlood Risk Categories (48h Ahead):")
print(df["flood_risk_target"].value_counts(normalize=True).round(3))

print("\n=== 4. PHYSICAL CORRELATION CHECKS ===")
# Temperature and Humidity should be inversely correlated (hotter = usually drier air)
temp_hum_corr = df["temperature"].corr(df["humidity"])
print(f"Temp vs Humidity Correlation: {temp_hum_corr:.2f} (Expected: Negative)")

# Rain accumulation vs Soil Moisture (Remember soil_moisture is low when wet!)
soil_rain_corr = df["rain_accum_24h"].corr(df["soil_moisture"])
print(f"24h Rain Accum vs Raw Soil Moisture Correlation: {soil_rain_corr:.2f} (Expected: Strong Negative)")

```
